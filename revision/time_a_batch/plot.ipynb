```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

from itertools import cycle
from sklearn.linear_model import LinearRegression
```

```python
resize = 1
plt.rcParams.update({
    "figure.figsize": (6.4*resize, 4.0*resize), # (6.4, 4.8)[4:3] -> (6.4, 4.0)[8:5]
    "font.sans-serif": ["Helvetica", "Nimbus Sans", "Arial", "DejaVu Sans"],
})
```

```python
npy_path = "./npy"
```

```python
x_base_2 = np.arange(2, 501, 2)
x_base_5 = np.arange(5, 501, 5)
x_base_10 = np.arange(10, 501, 10)

base_2 = []
base_5 = []
base_10 = []

base_2.append(np.load(f"{npy_path}/timePerBatch_cifar100_d18_2_501_2.npy"))
base_2.append(np.load(f"{npy_path}/timePerBatch_imagenet_d18_2_501_2.npy"))
base_2.append(np.load(f"{npy_path}/timePerBatch_imagenet_d34_2_501_2.npy"))

base_5.append(np.load(f"{npy_path}/timePerBatch_cifar100_d18_5_501_5.npy"))
base_5.append(np.load(f"{npy_path}/timePerBatch_imagenet_d18_5_501_5.npy"))
base_5.append(np.load(f"{npy_path}/timePerBatch_imagenet_d34_5_501_5.npy"))

base_10.append(np.load(f"{npy_path}/timePerBatch_cifar100_d18_10_501_10.npy"))
base_10.append(np.load(f"{npy_path}/timePerBatch_imagenet_d18_10_501_10.npy"))
base_10.append(np.load(f"{npy_path}/timePerBatch_imagenet_d34_10_501_10.npy"))
```

```python
pred_b2 = []
pred_b5 = []
pred_b10 = []

for i in base_2:
    reg_model = LinearRegression().fit(x_base_2.reshape(-1, 1), i)
    pred_b2.append(reg_model.predict(x_base_2.reshape(-1, 1)))
    print("Coefficient:", reg_model.coef_, "Intercept:", reg_model.intercept_)

print("====")

for i in base_5:
    reg_model = LinearRegression().fit(x_base_5.reshape(-1, 1), i)
    pred_b5.append(reg_model.predict(x_base_5.reshape(-1, 1)))
    print("Coefficient:", reg_model.coef_, "Intercept:", reg_model.intercept_)

print("====")

for i in base_10:
    reg_model = LinearRegression().fit(x_base_10.reshape(-1, 1), i)
    pred_b10.append(reg_model.predict(x_base_10.reshape(-1, 1)))
    print("Coefficient:", reg_model.coef_, "Intercept:", reg_model.intercept_)
```

```python
color_iter = cycle(mcolors.TABLEAU_COLORS.values())
label = ["cifar100 d18", "imagenet d18", "imagenet d34"]

plt.figure()

for lb, dot, pred in zip(label, base_2, pred_b2):
    plt.scatter(x_base_2, dot, s=5, label=lb, color=next(color_iter))
    plt.plot(x_base_2, pred, "--", label=lb, color=next(color_iter))

plt.xlabel('Batch Size')
plt.ylabel('Time (sec)')
plt.legend()
```

```python
color_iter = cycle(mcolors.TABLEAU_COLORS.values())
label = ["cifar100 d18", "imagenet d18", "imagenet d34"]

plt.figure()

for lb, dot, pred in zip(label, base_5, pred_b5):
    plt.scatter(x_base_5, dot, s=5, label=lb, color=next(color_iter))
    plt.plot(x_base_5, pred, "--", label=lb, color=next(color_iter))

plt.xlabel('Batch Size')
plt.ylabel('Time (sec)')
plt.legend()
```
